```python
import pandas as pd
from chembl_webresource_client.new_client import new_client

fields = ["molecule_chembl_id", "canonical_smiles", "standard_type",
          "standard_relation", "standard_value", "standard_units",
          "assay_chembl_id", "document_chembl_id", "target_chembl_id"]

res = new_client.activity.filter(
    target_chembl_id="CHEMBL350", standard_type="MIC"
).only(fields)

df = pd.DataFrame(list(res))   # can take several minutes
out = "/content/drive/MyDrive/kpneu_project/data/raw/chembl350_mic_activities.csv"
df.to_csv(out, index=False)

print(df.shape)
print(list(df.columns))
```

```python
assays = new_client.assay.filter(target_chembl_id="CHEMBL350")
adf = pd.DataFrame(list(assays))
print(adf.shape)
print(list(adf.columns))
```

```python
BASE = '/content/drive/MyDrive/kpneu_project'

keep = [c for c in ["assay_chembl_id", "assay_type", "assay_description",
                    "assay_organism", "assay_strain", "assay_cell_type"] if c in adf.columns]
adf = adf[keep].drop_duplicates("assay_chembl_id")
adf.to_csv(f"{BASE}/data/raw/chembl350_assays.csv", index=False)

print(adf.shape)
if 'assay_strain' in adf.columns:
    print(adf['assay_strain'].value_counts(dropna=False).head(15))
print(adf['assay_type'].value_counts(dropna=False))
```
